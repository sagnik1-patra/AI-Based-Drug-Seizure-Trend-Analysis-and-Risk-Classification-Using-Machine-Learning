```python
import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
import joblib

from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Drug Seizure Trend Analysis and Risk Classification Using Machine Learning"
)

CSV_PATH = BASE_DIR / "RS_Session_266_AU_242_C_iii.csv"

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

YEARS = ["2019", "2020", "2021", "2022", "2023"]

CLASS_NAMES = [
    "Low",
    "Medium",
    "High"
]

print("=" * 65)
print("AI-BASED DRUG SEIZURE ANALYSIS")
print("=" * 65)

print("Project folder:", BASE_DIR)

# ============================================================
# 2. LOAD DATASET
# ============================================================

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.astype(str).str.strip()

print("\nOriginal dataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst five rows:")
print(df.head())

# ============================================================
# 3. CLEAN DATASET
# ============================================================

drug_col = "Drug (In Kg) / Year"
serial_col = "Sl. No."

required_columns = [drug_col, serial_col] + YEARS

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )

# Keep only rows with a numeric serial number.
# This removes total and grand-total rows.

df[serial_col] = pd.to_numeric(
    df[serial_col],
    errors="coerce"
)

df = df.dropna(subset=[serial_col]).copy()

df["Drug"] = (
    df[drug_col]
    .astype(str)
    .str.strip()
)

# Exclude records measured in units other than kilograms.
# Tablets (In Kg) remains included.

exclude_pattern = (
    r"\(In No\.\)|"
    r"\(In Ampules\)|"
    r"\(In Bottles\)"
)

df = df[
    ~df["Drug"].str.contains(
        exclude_pattern,
        case=False,
        regex=True,
        na=False
    )
].copy()

for year in YEARS:
    df[year] = pd.to_numeric(
        df[year],
        errors="coerce"
    )

df[YEARS] = (
    df[YEARS]
    .fillna(0)
    .clip(lower=0)
)

df = df.drop_duplicates(
    subset=["Drug"],
    keep="first"
)

df = df.reset_index(drop=True)

df.to_csv(
    BASE_DIR / "cleaned_drug_seizure_data.csv",
    index=False
)

print("\nCleaned dataset shape:", df.shape)

# ============================================================
# 4. CONVERT WIDE DATA TO LONG FORMAT
# ============================================================

long_df = df.melt(
    id_vars=["Drug"],
    value_vars=YEARS,
    var_name="Year",
    value_name="Quantity_Kg"
)

long_df["Year"] = long_df["Year"].astype(int)

long_df = long_df.sort_values(
    ["Drug", "Year"]
).reset_index(drop=True)

long_df.to_csv(
    BASE_DIR / "drug_seizure_long_format.csv",
    index=False
)

print("\nLong-format dataset:")
print(long_df.head(10))

# ============================================================
# 5. FEATURE ENGINEERING
# ============================================================

# Use only historical information to predict
# the current year's seizure-volume category.

long_df["Previous_Quantity"] = (
    long_df.groupby("Drug")["Quantity_Kg"]
    .shift(1)
)

long_df["Two_Years_Ago"] = (
    long_df.groupby("Drug")["Quantity_Kg"]
    .shift(2)
)

# Use previous-year quantity and the preceding
# two-year trend. No current-year quantity is
# included in the predictor features.

long_df["Previous_Log_Quantity"] = np.log1p(
    long_df["Previous_Quantity"]
)

long_df["Previous_Change"] = (
    long_df["Previous_Quantity"]
    - long_df["Two_Years_Ago"]
)

long_df["Previous_Log_Change"] = (
    np.log1p(long_df["Previous_Quantity"])
    - np.log1p(long_df["Two_Years_Ago"])
)

# 2021-2023 have the required historical features.

model_df = long_df[
    long_df["Year"] >= 2021
].copy()

FEATURE_COLUMNS = [
    "Previous_Log_Quantity",
    "Previous_Log_Change",
    "Year"
]

model_df = model_df.dropna(
    subset=FEATURE_COLUMNS + ["Quantity_Kg"]
).copy()

# ============================================================
# 6. CHRONOLOGICAL TRAIN / TEST SPLIT
# ============================================================

train_df = model_df[
    model_df["Year"] < 2023
].copy()

test_df = model_df[
    model_df["Year"] == 2023
].copy()

if train_df.empty or test_df.empty:
    raise ValueError(
        "Insufficient chronological training/test data."
    )

# ============================================================
# 7. DEFINE SEIZURE-VOLUME CATEGORIES
# ============================================================

# Determine thresholds from training data only.
# This avoids learning category boundaries from
# the 2023 test observations.

q1, q2 = np.quantile(
    train_df["Quantity_Kg"],
    [1 / 3, 2 / 3]
)

if q1 >= q2:
    raise ValueError(
        "Cannot establish three distinct quantity classes."
    )

def classify_quantity(quantity):
    if quantity <= q1:
        return 0
    elif quantity <= q2:
        return 1
    return 2

train_df["Risk_Class"] = (
    train_df["Quantity_Kg"]
    .apply(classify_quantity)
)

test_df["Risk_Class"] = (
    test_df["Quantity_Kg"]
    .apply(classify_quantity)
)

model_df["Risk_Class"] = (
    model_df["Quantity_Kg"]
    .apply(classify_quantity)
)

model_df["Risk_Label"] = (
    model_df["Risk_Class"]
    .map(dict(enumerate(CLASS_NAMES)))
)

model_df.to_csv(
    BASE_DIR / "drug_risk_classification_data.csv",
    index=False
)

print("\nSeizure-volume thresholds (kg):")
print("Low: <=", round(float(q1), 3))
print("Medium: <=", round(float(q2), 3))
print("High: >", round(float(q2), 3))

# ============================================================
# 8. PREPARE TRAINING ARRAYS
# ============================================================

X_train = train_df[FEATURE_COLUMNS]
X_test = test_df[FEATURE_COLUMNS]

y_train = train_df["Risk_Class"].astype(int)
y_test = test_df["Risk_Class"].astype(int)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# 9. RANDOM FOREST CLASSIFIER
# ============================================================

rf_pipeline = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        RandomForestClassifier(
            n_estimators=150,
            max_depth=5,
            min_samples_leaf=2,
            random_state=SEED,
            class_weight="balanced"
        )
    )
])

rf_pipeline.fit(X_train, y_train)

rf_predictions = rf_pipeline.predict(X_test)

# Save the complete preprocessing + model pipeline.

pkl_path = BASE_DIR / "drug_seizure_model.pkl"

joblib.dump(
    {
        "pipeline": rf_pipeline,
        "features": FEATURE_COLUMNS,
        "thresholds_kg": [
            float(q1),
            float(q2)
        ],
        "classes": CLASS_NAMES,
        "model_type": "RandomForestClassifier",
        "prediction_target": "seizure_volume_category"
    },
    pkl_path
)

print("\nRandom Forest model saved:")
print(pkl_path)

# ============================================================
# 10. NEURAL NETWORK MODEL
# ============================================================

# Neural network uses scaled numeric features.

nn_scaler = StandardScaler()

X_train_nn = nn_scaler.fit_transform(
    X_train
).astype("float32")

X_test_nn = nn_scaler.transform(
    X_test
).astype("float32")

nn_model = Sequential([
    Input(shape=(len(FEATURE_COLUMNS),)),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.15),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        3,
        activation="softmax"
    )
])

nn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

nn_model.summary()

early_stop = EarlyStopping(
    monitor="loss",
    patience=15,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_nn,
    y_train.to_numpy(),
    epochs=100,
    batch_size=8,
    callbacks=[early_stop],
    verbose=1,
    shuffle=True
)

nn_probabilities = nn_model.predict(
    X_test_nn,
    verbose=0
)

nn_predictions = np.argmax(
    nn_probabilities,
    axis=1
)

# Save actual trained neural network in H5 format.

h5_path = BASE_DIR / "drug_seizure_model.h5"

nn_model.save(str(h5_path))

# Save neural network scaler separately.

joblib.dump(
    {
        "scaler": nn_scaler,
        "features": FEATURE_COLUMNS,
        "thresholds_kg": [
            float(q1),
            float(q2)
        ],
        "classes": CLASS_NAMES
    },
    BASE_DIR / "neural_network_preprocessor.pkl"
)

print("\nNeural network saved:")
print(h5_path)

# ============================================================
# 11. MODEL EVALUATION
# ============================================================

def evaluate_model(
    model_name,
    y_true,
    y_pred
):
    return {
        "model": model_name,
        "accuracy": float(
            accuracy_score(y_true, y_pred)
        ),
        "precision_weighted": float(
            precision_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            )
        ),
        "recall_weighted": float(
            recall_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            )
        ),
        "f1_weighted": float(
            f1_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            )
        )
    }

rf_metrics = evaluate_model(
    "Random Forest",
    y_test,
    rf_predictions
)

nn_metrics = evaluate_model(
    "Neural Network",
    y_test,
    nn_predictions
)

comparison_df = pd.DataFrame([
    rf_metrics,
    nn_metrics
])

comparison_df.to_csv(
    BASE_DIR / "model_comparison.csv",
    index=False
)

print("\nMODEL COMPARISON")
print(comparison_df.to_string(index=False))

print("\nRANDOM FOREST CLASSIFICATION REPORT")
print(
    classification_report(
        y_test,
        rf_predictions,
        labels=[0, 1, 2],
        target_names=CLASS_NAMES,
        zero_division=0
    )
)

print("\nNEURAL NETWORK CLASSIFICATION REPORT")
print(
    classification_report(
        y_test,
        nn_predictions,
        labels=[0, 1, 2],
        target_names=CLASS_NAMES,
        zero_division=0
    )
)

# ============================================================
# 12. SAVE PREDICTION RESULTS
# ============================================================

results_df = test_df[
    [
        "Drug",
        "Year",
        "Quantity_Kg",
        "Previous_Quantity"
    ]
].copy()

results_df["Actual_Class"] = y_test.to_numpy()

results_df["RF_Predicted_Class"] = (
    rf_predictions
)

results_df["NN_Predicted_Class"] = (
    nn_predictions
)

results_df["Actual_Label"] = [
    CLASS_NAMES[i]
    for i in results_df["Actual_Class"]
]

results_df["RF_Predicted_Label"] = [
    CLASS_NAMES[i]
    for i in rf_predictions
]

results_df["NN_Predicted_Label"] = [
    CLASS_NAMES[i]
    for i in nn_predictions
]

results_df.to_csv(
    BASE_DIR / "classification_results.csv",
    index=False
)

print("\nClassification results:")
print(results_df.head())

# ============================================================
# 13. SAVE MODEL CONFIGURATION TO YAML
# ============================================================

config = {
    "project_name": (
        "AI-Based Drug Seizure Trend Analysis "
        "and Risk Classification"
    ),

    "dataset": CSV_PATH.name,

    "dataset_information": {
        "original_records": 30,
        "usable_drug_records": int(len(df)),
        "measurement_unit": "kg",
        "historical_years": YEARS,
        "training_years": [2021, 2022],
        "testing_year": 2023
    },

    "features": FEATURE_COLUMNS,

    "target": "Risk_Class",

    "classification": {
        "method": "training_quantiles",
        "classes": CLASS_NAMES,
        "low_threshold_kg": float(q1),
        "medium_threshold_kg": float(q2),
        "interpretation": (
            "Relative seizure volume, "
            "not actual crime or health risk"
        )
    },

    "random_forest": {
        "n_estimators": 150,
        "max_depth": 5,
        "min_samples_leaf": 2,
        "random_state": SEED
    },

    "neural_network": {
        "hidden_layers": [32, 16],
        "output_units": 3,
        "activation": "relu",
        "output_activation": "softmax",
        "optimizer": "adam",
        "loss": "sparse_categorical_crossentropy",
        "epochs_max": 100,
        "batch_size": 8
    },

    "output_files": {
        "h5": "drug_seizure_model.h5",
        "pkl": "drug_seizure_model.pkl",
        "yaml": "model_config.yaml",
        "json": "model_metrics.json"
    }
}

yaml_path = BASE_DIR / "model_config.yaml"

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("\nYAML configuration saved:")
print(yaml_path)

# ============================================================
# 14. SAVE METRICS TO JSON
# ============================================================

metrics_data = {
    "project": (
        "AI-Based Drug Seizure Trend Analysis "
        "and Risk Classification"
    ),

    "training_samples": int(len(X_train)),
    "testing_samples": int(len(X_test)),

    "training_years": [2021, 2022],
    "testing_year": 2023,

    "class_distribution_training": {
        CLASS_NAMES[i]: int((y_train == i).sum())
        for i in range(3)
    },

    "class_distribution_testing": {
        CLASS_NAMES[i]: int((y_test == i).sum())
        for i in range(3)
    },

    "thresholds_kg": {
        "low": float(q1),
        "medium": float(q2)
    },

    "random_forest": rf_metrics,
    "neural_network": nn_metrics,

    "random_forest_confusion_matrix": (
        confusion_matrix(
            y_test,
            rf_predictions,
            labels=[0, 1, 2]
        ).tolist()
    ),

    "neural_network_confusion_matrix": (
        confusion_matrix(
            y_test,
            nn_predictions,
            labels=[0, 1, 2]
        ).tolist()
    ),

    "evaluation_note": (
        "Small exploratory dataset. "
        "Results should not be interpreted as "
        "validated operational risk predictions."
    )
}

json_path = BASE_DIR / "model_metrics.json"

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metrics_data,
        file,
        indent=4
    )

print("\nJSON metrics saved:")
print(json_path)

# ============================================================
# 15. ACCURACY COMPARISON GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    data=comparison_df,
    x="model",
    y="accuracy"
)

plt.title("Model Accuracy Comparison")
plt.xlabel("Machine Learning Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "accuracy_graph.png",
    dpi=300
)

plt.show()

# ============================================================
# 16. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    rf_predictions,
    labels=[0, 1, 2]
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES
)

plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()

plt.savefig(
    BASE_DIR / "confusion_matrix.png",
    dpi=300
)

plt.show()

# ============================================================
# 17. HISTORICAL SEIZURE TREND GRAPH
# ============================================================

yearly_totals = (
    long_df.groupby("Year")["Quantity_Kg"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(9, 5))

plt.plot(
    yearly_totals["Year"],
    yearly_totals["Quantity_Kg"],
    marker="o",
    linewidth=2
)

plt.title("Historical Drug Seizure Trend")
plt.xlabel("Year")
plt.ylabel("Total Seizure Quantity (Kg)")
plt.xticks(yearly_totals["Year"])
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "seizure_trend_graph.png",
    dpi=300
)

plt.show()

# ============================================================
# 18. FEATURE IMPORTANCE
# ============================================================

rf_classifier = rf_pipeline.named_steps[
    "classifier"
]

importance_df = pd.DataFrame({
    "Feature": FEATURE_COLUMNS,
    "Importance": rf_classifier.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

importance_df.to_csv(
    BASE_DIR / "feature_importance.csv",
    index=False
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature"
)

plt.title("Random Forest Feature Importance")
plt.tight_layout()

plt.savefig(
    BASE_DIR / "feature_importance_graph.png",
    dpi=300
)

plt.show()

# ============================================================
# 19. NEURAL NETWORK TRAINING GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.title("Neural Network Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "nn_training_accuracy.png",
    dpi=300
)

plt.show()

# ============================================================
# 20. VERIFY SAVED FILES
# ============================================================

required_files = [
    "drug_seizure_model.h5",
    "drug_seizure_model.pkl",
    "neural_network_preprocessor.pkl",
    "model_config.yaml",
    "model_metrics.json"
]

print("\n" + "=" * 65)
print("FINAL OUTPUT VERIFICATION")
print("=" * 65)

for filename in required_files:
    filepath = BASE_DIR / filename

    if filepath.exists():
        print(
            "[SAVED]",
            filename,
            "-",
            filepath.stat().st_size,
            "bytes"
        )
    else:
        print("[MISSING]", filename)

# ============================================================
# 21. TEST LOADING SAVED MODELS
# ============================================================

loaded_rf = joblib.load(
    BASE_DIR / "drug_seizure_model.pkl"
)

loaded_nn = tf.keras.models.load_model(
    str(BASE_DIR / "drug_seizure_model.h5"),
    compile=False
)

loaded_rf_predictions = (
    loaded_rf["pipeline"].predict(X_test)
)

loaded_nn_predictions = np.argmax(
    loaded_nn.predict(
        X_test_nn,
        verbose=0
    ),
    axis=1
)

assert np.array_equal(
    loaded_rf_predictions,
    rf_predictions
)

assert np.array_equal(
    loaded_nn_predictions,
    nn_predictions
)

print("\nBoth saved models loaded successfully.")

print("\nPROJECT COMPLETED SUCCESSFULLY")
print("All files saved in:")
print(BASE_DIR)
```
