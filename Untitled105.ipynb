```python
import warnings
warnings.filterwarnings("ignore")

import json
import copy
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Drug Seizure Trend Analysis and Risk Classification Using Machine Learning"
)

CSV_PATH = (
    BASE_DIR / "RS_Session_266_AU_242_C_iii.csv"
)

BASE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

sns.set_theme(style="whitegrid")

YEARS = [
    "2019",
    "2020",
    "2021",
    "2022",
    "2023"
]

CLASS_NAMES = [
    "Low",
    "Medium",
    "High"
]

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

def save_show(filename):
    path = BASE_DIR / filename
    plt.tight_layout()
    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()
    print("[SAVED]", path.name)

def metrics(name, actual, predicted):
    return {
        "Model": name,
        "Accuracy": float(
            accuracy_score(actual, predicted)
        ),
        "Precision": float(
            precision_score(
                actual,
                predicted,
                average="weighted",
                zero_division=0
            )
        ),
        "Recall": float(
            recall_score(
                actual,
                predicted,
                average="weighted",
                zero_division=0
            )
        ),
        "F1_Score": float(
            f1_score(
                actual,
                predicted,
                average="weighted",
                zero_division=0
            )
        )
    }

# ============================================================
# 2. LOAD AND CLEAN DATA
# ============================================================

df = pd.read_csv(CSV_PATH)

df.columns = (
    df.columns.astype(str).str.strip()
)

DRUG_COL = "Drug (In Kg) / Year"
SERIAL_COL = "Sl. No."

required = [DRUG_COL, SERIAL_COL] + YEARS

missing = [
    col for col in required
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing dataset columns: {missing}"
    )

df[SERIAL_COL] = pd.to_numeric(
    df[SERIAL_COL],
    errors="coerce"
)

# Exclude summary rows.
df = df.dropna(
    subset=[SERIAL_COL]
).copy()

df["Drug"] = (
    df[DRUG_COL]
    .astype(str)
    .str.strip()
)

# Exclude non-kilogram measurements.
exclude_pattern = (
    r"\(In No\.\)|"
    r"\(In Ampules\)|"
    r"\(In Bottles\)"
)

df = df[
    ~df["Drug"].str.contains(
        exclude_pattern,
        case=False,
        regex=True,
        na=False
    )
].copy()

for year in YEARS:
    df[year] = pd.to_numeric(
        df[year],
        errors="coerce"
    )

df = df.dropna(
    subset=YEARS
).copy()

df[YEARS] = df[YEARS].clip(
    lower=0
)

df = df.drop_duplicates(
    subset=["Drug"]
).reset_index(drop=True)

df.to_csv(
    BASE_DIR / "ais_cleaned_dataset.csv",
    index=False
)

print("Cleaned records:", len(df))

# ============================================================
# 3. TRANSFORM TO LONG FORMAT
# ============================================================

long_df = df.melt(
    id_vars=["Drug"],
    value_vars=YEARS,
    var_name="Year",
    value_name="Quantity_Kg"
)

long_df["Year"] = (
    long_df["Year"].astype(int)
)

long_df = long_df.sort_values(
    ["Drug", "Year"]
).reset_index(drop=True)

long_df.to_csv(
    BASE_DIR / "ais_long_format.csv",
    index=False
)

# ============================================================
# 4. FEATURE ENGINEERING
# ============================================================

long_df["Previous_Quantity"] = (
    long_df.groupby("Drug")["Quantity_Kg"]
    .shift(1)
)

long_df["Two_Years_Ago"] = (
    long_df.groupby("Drug")["Quantity_Kg"]
    .shift(2)
)

long_df["Previous_Log_Quantity"] = (
    np.log1p(
        long_df["Previous_Quantity"]
    )
)

long_df["Previous_Log_Change"] = (
    np.log1p(
        long_df["Previous_Quantity"]
    )
    -
    np.log1p(
        long_df["Two_Years_Ago"]
    )
)

long_df["Previous_Growth"] = (
    (
        long_df["Previous_Quantity"]
        -
        long_df["Two_Years_Ago"]
    )
    /
    (
        long_df["Two_Years_Ago"] + 1
    )
)

FEATURES = [
    "Previous_Log_Quantity",
    "Previous_Log_Change",
    "Previous_Growth",
    "Year"
]

model_df = long_df[
    long_df["Year"] >= 2021
].dropna(
    subset=FEATURES + ["Quantity_Kg"]
).copy()

# ============================================================
# 5. TRAIN / TEST SPLIT
# ============================================================

train_df = model_df[
    model_df["Year"] <= 2022
].copy()

test_df = model_df[
    model_df["Year"] == 2023
].copy()

if train_df.empty or test_df.empty:
    raise ValueError(
        "Insufficient training or testing data."
    )

# Fit thresholds using training data only.
LOW_THRESHOLD, HIGH_THRESHOLD = np.quantile(
    train_df["Quantity_Kg"],
    [1/3, 2/3]
)

if LOW_THRESHOLD >= HIGH_THRESHOLD:
    raise ValueError(
        "Cannot construct three distinct classes."
    )

def classify_quantity(quantity):
    if quantity <= LOW_THRESHOLD:
        return 0
    elif quantity <= HIGH_THRESHOLD:
        return 1
    return 2

train_df["Risk_Class"] = (
    train_df["Quantity_Kg"]
    .apply(classify_quantity)
)

test_df["Risk_Class"] = (
    test_df["Quantity_Kg"]
    .apply(classify_quantity)
)

model_df["Risk_Class"] = (
    model_df["Quantity_Kg"]
    .apply(classify_quantity)
)

X_train = train_df[FEATURES]
y_train = train_df["Risk_Class"]

X_test = test_df[FEATURES]
y_test = test_df["Risk_Class"]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# 6. ARTIFICIAL IMMUNE SYSTEM
# ============================================================
# CLONALG-inspired optimization:
# - Each antibody represents a feature mask and RF parameters.
# - Affinity is measured using a 2022 validation set.
# - Higher-affinity antibodies are cloned.
# - Clones undergo mutation.
# - Low-affinity antibodies are replaced.
#
# Important: 2023 is never used during optimization.
# ============================================================

POPULATION_SIZE = 12
GENERATIONS = 15
ELITE_COUNT = 4
CLONES_PER_ELITE = 4
RANDOM_REPLACEMENTS = 3

# Use 2021 for AIS training and 2022 for validation.
ais_train_df = train_df[
    train_df["Year"] == 2021
].copy()

ais_val_df = train_df[
    train_df["Year"] == 2022
].copy()

if ais_train_df.empty or ais_val_df.empty:
    raise ValueError(
        "AIS validation years are unavailable."
    )

# The Year feature is constant within a year.
# Therefore, exclude it from AIS feature selection.
# This avoids selecting an uninformative feature.

AIS_FEATURES = [
    "Previous_Log_Quantity",
    "Previous_Log_Change",
    "Previous_Growth"
]

# NOTE:
# The 2021 samples use 2019 and 2020 history,
# while 2022 samples use 2020 and 2021 history.

X_ais_train = ais_train_df[AIS_FEATURES]
y_ais_train = ais_train_df["Risk_Class"]

X_ais_val = ais_val_df[AIS_FEATURES]
y_ais_val = ais_val_df["Risk_Class"]

def create_antibody():

    return {
        "mask": np.random.randint(
            0,
            2,
            size=len(AIS_FEATURES)
        ).tolist(),

        "n_estimators": random.randint(
            40,
            160
        ),

        "max_depth": random.randint(
            2,
            8
        ),

        "min_samples_leaf": random.randint(
            1,
            5
        )
    }

def selected_features(antibody):

    features = [
        feature
        for feature, selected in zip(
            AIS_FEATURES,
            antibody["mask"]
        )
        if selected == 1
    ]

    # Ensure at least one selected feature.
    if not features:
        features = [
            AIS_FEATURES[0]
        ]

    return features

def build_rf(antibody):

    return RandomForestClassifier(
        n_estimators=antibody["n_estimators"],
        max_depth=antibody["max_depth"],
        min_samples_leaf=antibody["min_samples_leaf"],
        class_weight="balanced",
        random_state=SEED,
        n_jobs=-1
    )

fitness_cache = {}

def antibody_fitness(antibody):

    features = selected_features(
        antibody
    )

    key = (
        tuple(features),
        antibody["n_estimators"],
        antibody["max_depth"],
        antibody["min_samples_leaf"]
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = build_rf(antibody)

    model.fit(
        X_ais_train[features],
        y_ais_train
    )

    predictions = model.predict(
        X_ais_val[features]
    )

    score = f1_score(
        y_ais_val,
        predictions,
        average="macro",
        zero_division=0
    )

    # Small feature-count penalty.
    penalty = 0.005 * len(features)

    fitness = float(score - penalty)

    fitness_cache[key] = fitness

    return fitness

def mutate_antibody(
    antibody,
    mutation_rate=0.25
):

    mutated = copy.deepcopy(
        antibody
    )

    for i in range(len(mutated["mask"])):

        if random.random() < mutation_rate:

            mutated["mask"][i] = (
                1 - mutated["mask"][i]
            )

    if random.random() < mutation_rate:

        mutated["n_estimators"] += random.randint(
            -25,
            25
        )

    if random.random() < mutation_rate:

        mutated["max_depth"] += random.randint(
            -2,
            2
        )

    if random.random() < mutation_rate:

        mutated["min_samples_leaf"] += random.choice(
            [-1, 1]
        )

    mutated["n_estimators"] = int(
        np.clip(
            mutated["n_estimators"],
            30,
            200
        )
    )

    mutated["max_depth"] = int(
        np.clip(
            mutated["max_depth"],
            2,
            10
        )
    )

    mutated["min_samples_leaf"] = int(
        np.clip(
            mutated["min_samples_leaf"],
            1,
            6
        )
    )

    return mutated

# ============================================================
# 7. RUN AIS OPTIMIZATION
# ============================================================

print("\nStarting AIS optimization...")

population = [
    create_antibody()
    for _ in range(POPULATION_SIZE)
]

best_antibody = None
best_fitness = -float("inf")

optimization_history = []

for generation in range(GENERATIONS):

    scored = [
        (
            antibody_fitness(antibody),
            antibody
        )
        for antibody in population
    ]

    scored.sort(
        key=lambda item: item[0],
        reverse=True
    )

    generation_best_score = scored[0][0]

    if generation_best_score > best_fitness:

        best_fitness = generation_best_score

        best_antibody = copy.deepcopy(
            scored[0][1]
        )

    optimization_history.append({
        "Generation": generation + 1,
        "Best_Fitness": best_fitness,
        "Generation_Best": generation_best_score
    })

    print(
        f"Generation {generation + 1}/{GENERATIONS}"
        f" | Best Fitness: {best_fitness:.4f}"
    )

    elites = [
        copy.deepcopy(item[1])
        for item in scored[:ELITE_COUNT]
    ]

    new_population = elites.copy()

    clones = []

    for rank, elite in enumerate(elites):

        # Higher-ranked antibodies get lower mutation.
        mutation_rate = (
            0.10 + 0.10 * rank
        )

        for _ in range(CLONES_PER_ELITE):

            clone = mutate_antibody(
                elite,
                mutation_rate=mutation_rate
            )

            clones.append(clone)

    clones.sort(
        key=antibody_fitness,
        reverse=True
    )

    available_slots = (
        POPULATION_SIZE
        - len(new_population)
        - RANDOM_REPLACEMENTS
    )

    new_population.extend(
        clones[:max(0, available_slots)]
    )

    for _ in range(RANDOM_REPLACEMENTS):
        new_population.append(
            create_antibody()
        )

    population = new_population[
        :POPULATION_SIZE
    ]

if best_antibody is None:
    raise RuntimeError(
        "AIS optimization failed."
    )

BEST_FEATURES = selected_features(
    best_antibody
)

print("\nAIS optimization completed.")
print("Best fitness:", best_fitness)
print("Selected features:", BEST_FEATURES)
print("Best antibody:", best_antibody)

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    BASE_DIR / "ais_optimization_history.csv",
    index=False
)

# ============================================================
# 8. AIS OPTIMIZATION GRAPH
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    history_df["Generation"],
    history_df["Best_Fitness"],
    marker="o",
    linewidth=2,
    label="Best AIS Fitness"
)

plt.title(
    "AIS Clonal Selection Optimization Progress"
)

plt.xlabel("Generation")
plt.ylabel("Fitness Score")

plt.legend()
plt.grid(alpha=0.3)

save_show("ais_optimization_graph.png")

# ============================================================
# 9. TRAIN BASELINE AND AIS MODELS
# ============================================================

baseline_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=SEED,
    class_weight="balanced",
    n_jobs=-1
)

baseline_rf.fit(
    X_train[AIS_FEATURES],
    y_train
)

baseline_pred = baseline_rf.predict(
    X_test[AIS_FEATURES]
)

ais_rf = build_rf(
    best_antibody
)

ais_rf.fit(
    X_train[BEST_FEATURES],
    y_train
)

ais_pred = ais_rf.predict(
    X_test[BEST_FEATURES]
)

# Additional comparison models.

decision_tree = DecisionTreeClassifier(
    max_depth=4,
    random_state=SEED
)

decision_tree.fit(
    X_train[AIS_FEATURES],
    y_train
)

dt_pred = decision_tree.predict(
    X_test[AIS_FEATURES]
)

logistic = Pipeline([
    ("scaler", StandardScaler()),
    (
        "model",
        LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=SEED
        )
    )
])

logistic.fit(
    X_train[AIS_FEATURES],
    y_train
)

lr_pred = logistic.predict(
    X_test[AIS_FEATURES]
)

# ============================================================
# 10. SAVE MODEL COMPARISON METRICS
# ============================================================

comparison_df = pd.DataFrame([
    metrics(
        "AIS Random Forest",
        y_test,
        ais_pred
    ),
    metrics(
        "Baseline Random Forest",
        y_test,
        baseline_pred
    ),
    metrics(
        "Decision Tree",
        y_test,
        dt_pred
    ),
    metrics(
        "Logistic Regression",
        y_test,
        lr_pred
    )
])

comparison_df.to_csv(
    BASE_DIR / "ais_comparison.csv",
    index=False
)

comparison_df.to_csv(
    BASE_DIR / "ais_accuracy_metrics.csv",
    index=False
)

print("\nMODEL COMPARISON")
print(comparison_df.to_string(index=False))

# ============================================================
# 11. AIS ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="Accuracy",
    hue="Model",
    palette="viridis",
    legend=False
)

plt.title(
    "AIS vs Baseline Model Accuracy"
)

plt.xlabel("Machine Learning Model")
plt.ylabel("Accuracy")

plt.ylim(0, 1.1)

plt.xticks(
    rotation=15,
    ha="right"
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.3f",
        padding=3
    )

save_show("ais_accuracy_graph.png")

# ============================================================
# 12. AIS COMPARISON GRAPH
# ============================================================

comparison_long = comparison_df.melt(
    id_vars="Model",
    value_vars=[
        "Accuracy",
        "Precision",
        "Recall",
        "F1_Score"
    ],
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=comparison_long,
    x="Model",
    y="Score",
    hue="Metric"
)

plt.title(
    "AIS Comparative Model Performance"
)

plt.xlabel("Model")
plt.ylabel("Score")

plt.ylim(0, 1.05)

plt.xticks(
    rotation=15,
    ha="right"
)

save_show("ais_comparison_graph.png")

# ============================================================
# 13. AIS CORRELATION HEATMAP
# ============================================================

heatmap_features = list(
    dict.fromkeys(
        BEST_FEATURES + ["Quantity_Kg"]
    )
)

correlation_matrix = train_df[
    heatmap_features
].corr(
    numeric_only=True
)

correlation_matrix.to_csv(
    BASE_DIR / "ais_correlation_matrix.csv"
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    square=True
)

plt.title(
    "AIS Selected Feature Correlation Heatmap"
)

save_show("ais_heatmap.png")

# ============================================================
# 14. SAVE AIS RESULT CSV
# ============================================================

result_df = test_df[
    [
        "Drug",
        "Year",
        "Quantity_Kg",
        "Previous_Quantity"
    ]
].copy()

result_df["Actual_Class"] = (
    y_test.to_numpy()
)

result_df["AIS_Predicted_Class"] = (
    ais_pred
)

result_df["Actual_Risk"] = [
    CLASS_NAMES[int(value)]
    for value in y_test
]

result_df["AIS_Predicted_Risk"] = [
    CLASS_NAMES[int(value)]
    for value in ais_pred
]

result_df["Correct_Prediction"] = (
    result_df["Actual_Class"]
    ==
    result_df["AIS_Predicted_Class"]
)

result_df.to_csv(
    BASE_DIR / "ais_result.csv",
    index=False
)

print("\nAIS RESULT CSV")
print(result_df.head(10))

# ============================================================
# 15. AIS RESULT GRAPH
# ============================================================

result_plot = result_df.sort_values(
    "Drug"
).reset_index(drop=True)

x = np.arange(
    len(result_plot)
)

plt.figure(figsize=(15, 7))

plt.plot(
    x,
    result_plot["Actual_Class"],
    marker="o",
    linewidth=2,
    label="Actual"
)

plt.plot(
    x,
    result_plot["AIS_Predicted_Class"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Predicted"
)

plt.xticks(
    x,
    result_plot["Drug"],
    rotation=75,
    ha="right",
    fontsize=8
)

plt.yticks(
    [0, 1, 2],
    CLASS_NAMES
)

plt.title(
    "AIS Actual vs Predicted Risk Classification (2023)"
)

plt.xlabel("Drug")
plt.ylabel("Seizure-Volume Category")

plt.legend()
plt.grid(alpha=0.3)

save_show("ais_result_graph.png")

# ============================================================
# 16. AIS CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    ais_pred,
    labels=[0, 1, 2]
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES
)

plt.title(
    "AIS Random Forest Confusion Matrix"
)

plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")

save_show("ais_confusion_matrix.png")

# ============================================================
# 17. GENERATE 2024 AIS PREDICTIONS
# ============================================================

# Use 2022 and 2023 as historical inputs.

future_df = df[
    ["Drug", "2022", "2023"]
].copy()

future_df = future_df.rename(
    columns={
        "2022": "Two_Years_Ago",
        "2023": "Previous_Quantity"
    }
)

future_df["Year"] = 2024

future_df["Previous_Log_Quantity"] = (
    np.log1p(
        future_df["Previous_Quantity"]
    )
)

future_df["Previous_Log_Change"] = (
    np.log1p(
        future_df["Previous_Quantity"]
    )
    -
    np.log1p(
        future_df["Two_Years_Ago"]
    )
)

future_df["Previous_Growth"] = (
    (
        future_df["Previous_Quantity"]
        -
        future_df["Two_Years_Ago"]
    )
    /
    (
        future_df["Two_Years_Ago"] + 1
    )
)

# Refit AIS-selected model on all available
# labeled years before generating 2024 predictions.

final_ais_model = build_rf(
    best_antibody
)

final_ais_model.fit(
    model_df[BEST_FEATURES],
    model_df["Risk_Class"]
)

future_pred = final_ais_model.predict(
    future_df[BEST_FEATURES]
)

future_proba = final_ais_model.predict_proba(
    future_df[BEST_FEATURES]
)

prediction_df = future_df[
    [
        "Drug",
        "Year",
        "Previous_Quantity"
    ]
].copy()

prediction_df["AIS_Predicted_Class"] = (
    future_pred
)

prediction_df["AIS_Predicted_Risk"] = [
    CLASS_NAMES[int(value)]
    for value in future_pred
]

for i, class_id in enumerate(
    final_ais_model.classes_
):
    prediction_df[
        f"Probability_{CLASS_NAMES[int(class_id)]}"
    ] = future_proba[:, i]

prediction_df["Confidence"] = (
    future_proba.max(axis=1)
)

prediction_df["Forecast_Type"] = (
    "Experimental AIS 2024 classification"
)

prediction_df.to_csv(
    BASE_DIR / "ais_prediction.csv",
    index=False
)

print("\nAIS PREDICTION CSV")
print(prediction_df.head(10))

# ============================================================
# 18. AIS PREDICTION GRAPH
# ============================================================

prediction_plot = prediction_df.sort_values(
    "Drug"
).reset_index(drop=True)

x = np.arange(
    len(prediction_plot)
)

colors = [
    "#2ca25f" if value == 0
    else "#feb24c" if value == 1
    else "#de2d26"
    for value in prediction_plot["AIS_Predicted_Class"]
]

plt.figure(figsize=(15, 7))

plt.bar(
    x,
    prediction_plot["AIS_Predicted_Class"],
    color=colors
)

plt.xticks(
    x,
    prediction_plot["Drug"],
    rotation=75,
    ha="right",
    fontsize=8
)

plt.yticks(
    [0, 1, 2],
    CLASS_NAMES
)

plt.ylim(-0.1, 2.5)

plt.title(
    "AIS Experimental Drug Seizure Predictions for 2024"
)

plt.xlabel("Drug")
plt.ylabel("Predicted Seizure-Volume Category")

save_show("ais_prediction_graph.png")

# ============================================================
# 19. AIS FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.DataFrame({
    "Feature": BEST_FEATURES,
    "Importance": final_ais_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=False
)

feature_importance.to_csv(
    BASE_DIR / "ais_feature_importance.csv",
    index=False
)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=feature_importance,
    x="Importance",
    y="Feature",
    hue="Feature",
    legend=False
)

plt.title(
    "AIS-Optimized Feature Importance"
)

save_show("ais_feature_importance_graph.png")

# ============================================================
# 20. SAVE AIS MODEL
# ============================================================

joblib.dump(
    {
        "model": final_ais_model,
        "selected_features": BEST_FEATURES,
        "best_antibody": best_antibody,
        "class_names": CLASS_NAMES,
        "low_threshold_kg": float(LOW_THRESHOLD),
        "high_threshold_kg": float(HIGH_THRESHOLD),
        "optimization_method": "CLONALG-inspired AIS"
    },
    BASE_DIR / "ais_model.pkl"
)

# ============================================================
# 21. SAVE AIS METRICS JSON
# ============================================================

ais_metrics = {
    "project": (
        "AI-Based Drug Seizure Trend Analysis "
        "and Risk Classification"
    ),
    "optimization": "Artificial Immune System",
    "algorithm": "CLONALG-inspired Clonal Selection",
    "generations": GENERATIONS,
    "population_size": POPULATION_SIZE,
    "best_fitness": float(best_fitness),
    "selected_features": BEST_FEATURES,
    "best_antibody": best_antibody,
    "training_samples": int(len(X_train)),
    "testing_samples": int(len(X_test)),
    "training_years": [2021, 2022],
    "testing_year": 2023,
    "classification_thresholds_kg": {
        "low": float(LOW_THRESHOLD),
        "medium": float(HIGH_THRESHOLD)
    },
    "models": {
        row["Model"]: {
            "accuracy": float(row["Accuracy"]),
            "precision": float(row["Precision"]),
            "recall": float(row["Recall"]),
            "f1_score": float(row["F1_Score"])
        }
        for _, row in comparison_df.iterrows()
    },
    "confusion_matrix": cm.tolist(),
    "evaluation_note": (
        "Exploratory analysis on a small dataset. "
        "The volume classes do not measure real-world "
        "crime or public-health risk."
    )
}

with open(
    BASE_DIR / "ais_metrics.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        ais_metrics,
        file,
        indent=4
    )

# ============================================================
# 22. SAVE AIS CONFIGURATION YAML
# ============================================================

ais_config = {
    "project": (
        "AI-Based Drug Seizure Trend Analysis "
        "and Risk Classification"
    ),
    "dataset": CSV_PATH.name,
    "optimization_method": "Artificial Immune System",
    "algorithm": "CLONALG-inspired",
    "population_size": POPULATION_SIZE,
    "generations": GENERATIONS,
    "elite_count": ELITE_COUNT,
    "clones_per_elite": CLONES_PER_ELITE,
    "selected_features": BEST_FEATURES,
    "best_antibody": best_antibody,
    "target": "Risk_Class",
    "class_names": CLASS_NAMES,
    "training_years": [2021, 2022],
    "testing_year": 2023,
    "forecast_year": 2024,
    "model_file": "ais_model.pkl"
}

with open(
    BASE_DIR / "ais_config.yaml",
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        ais_config,
        file,
        sort_keys=False
    )

# ============================================================
# 23. FINAL CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("AIS CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        ais_pred,
        labels=[0, 1, 2],
        target_names=CLASS_NAMES,
        zero_division=0
    )
)

print("\nAIS Accuracy:", accuracy_score(
    y_test,
    ais_pred
))

print("\nAIS F1 Score:", f1_score(
    y_test,
    ais_pred,
    average="weighted",
    zero_division=0
))

# ============================================================
# 24. VERIFY OUTPUT FILES
# ============================================================

output_files = [
    "ais_cleaned_dataset.csv",
    "ais_long_format.csv",
    "ais_optimization_history.csv",
    "ais_optimization_graph.png",
    "ais_accuracy_metrics.csv",
    "ais_accuracy_graph.png",
    "ais_comparison.csv",
    "ais_comparison_graph.png",
    "ais_correlation_matrix.csv",
    "ais_heatmap.png",
    "ais_result.csv",
    "ais_result_graph.png",
    "ais_confusion_matrix.png",
    "ais_prediction.csv",
    "ais_prediction_graph.png",
    "ais_feature_importance.csv",
    "ais_feature_importance_graph.png",
    "ais_model.pkl",
    "ais_metrics.json",
    "ais_config.yaml"
]

print("\n" + "=" * 70)
print("AIS OUTPUT VERIFICATION")
print("=" * 70)

for filename in output_files:

    filepath = BASE_DIR / filename

    if filepath.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\n" + "=" * 70)
print("AIS PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nAll files saved in:")
print(BASE_DIR)
```
