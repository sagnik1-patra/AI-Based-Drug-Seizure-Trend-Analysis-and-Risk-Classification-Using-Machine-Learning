```python
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.tree import DecisionTreeClassifier

from sklearn.linear_model import LogisticRegression

from sklearn.preprocessing import StandardScaler

from sklearn.pipeline import Pipeline

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ================================================================
# 1. PROJECT PATHS
# ================================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Drug Seizure Trend Analysis and Risk Classification Using Machine Learning"
)

CSV_PATH = (
    BASE_DIR / "RS_Session_266_AU_242_C_iii.csv"
)

BASE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_PATH}"
    )

SEED = 42

np.random.seed(SEED)

sns.set_theme(
    style="whitegrid",
    palette="deep"
)

CLASS_NAMES = [
    "Low",
    "Medium",
    "High"
]

YEARS = [
    "2019",
    "2020",
    "2021",
    "2022",
    "2023"
]

# ================================================================
# 2. HELPER FUNCTIONS
# ================================================================

def save_show(filename):
    """
    Save the current graph and display it.
    """
    filepath = BASE_DIR / filename

    plt.tight_layout()

    plt.savefig(
        filepath,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print(f"[SAVED] {filepath}")


def evaluate_model(name, actual, predicted):

    return {
        "Model": name,

        "Accuracy": accuracy_score(
            actual,
            predicted
        ),

        "Precision": precision_score(
            actual,
            predicted,
            average="weighted",
            zero_division=0
        ),

        "Recall": recall_score(
            actual,
            predicted,
            average="weighted",
            zero_division=0
        ),

        "F1_Score": f1_score(
            actual,
            predicted,
            average="weighted",
            zero_division=0
        )
    }


# ================================================================
# 3. LOAD DATASET
# ================================================================

print("=" * 75)
print("LOADING DRUG SEIZURE DATASET")
print("=" * 75)

df = pd.read_csv(
    CSV_PATH
)

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
)

print("\nDataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst five records:")
print(df.head())

# ================================================================
# 4. CLEAN DATASET
# ================================================================

DRUG_COLUMN = "Drug (In Kg) / Year"

SERIAL_COLUMN = "Sl. No."

required_columns = (
    [DRUG_COLUMN, SERIAL_COLUMN] + YEARS
)

missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

# Remove aggregate and summary rows.

df[SERIAL_COLUMN] = pd.to_numeric(
    df[SERIAL_COLUMN],
    errors="coerce"
)

df = df.dropna(
    subset=[SERIAL_COLUMN]
).copy()

df["Drug"] = (
    df[DRUG_COLUMN]
    .astype(str)
    .str.strip()
)

# Keep kilogram-based observations only.
# Exclude separately measured quantities.

exclude_pattern = (
    r"\(In No\.\)|"
    r"\(In Ampules\)|"
    r"\(In Bottles\)"
)

df = df[
    ~df["Drug"].str.contains(
        exclude_pattern,
        case=False,
        regex=True,
        na=False
    )
].copy()

for year in YEARS:

    df[year] = pd.to_numeric(
        df[year],
        errors="coerce"
    )

# Treat missing historical quantities as missing,
# not automatically as zero.

df = df.dropna(
    subset=YEARS
).copy()

df[YEARS] = df[YEARS].clip(
    lower=0
)

df = df.drop_duplicates(
    subset=["Drug"]
).reset_index(drop=True)

if df.empty:
    raise ValueError(
        "No usable kilogram-based records found."
    )

df.to_csv(
    BASE_DIR / "cleaned_dataset.csv",
    index=False
)

print("\nCleaned dataset shape:", df.shape)

# ================================================================
# 5. CONVERT DATA TO LONG FORMAT
# ================================================================

long_df = df.melt(
    id_vars=["Drug"],
    value_vars=YEARS,
    var_name="Year",
    value_name="Quantity_Kg"
)

long_df["Year"] = (
    long_df["Year"].astype(int)
)

long_df = long_df.sort_values(
    ["Drug", "Year"]
).reset_index(drop=True)

print("\nLong-format data:")
print(long_df.head(10))

# ================================================================
# 6. FEATURE ENGINEERING
# ================================================================

# Predict current-year volume category using
# only information from earlier years.

long_df["Previous_Quantity"] = (
    long_df.groupby("Drug")["Quantity_Kg"]
    .shift(1)
)

long_df["Two_Years_Ago"] = (
    long_df.groupby("Drug")["Quantity_Kg"]
    .shift(2)
)

long_df["Previous_Log_Quantity"] = (
    np.log1p(
        long_df["Previous_Quantity"]
    )
)

long_df["Previous_Log_Change"] = (
    np.log1p(
        long_df["Previous_Quantity"]
    )
    -
    np.log1p(
        long_df["Two_Years_Ago"]
    )
)

long_df["Previous_Growth"] = (
    (
        long_df["Previous_Quantity"]
        -
        long_df["Two_Years_Ago"]
    )
    /
    (
        long_df["Two_Years_Ago"] + 1
    )
)

FEATURE_COLUMNS = [
    "Previous_Log_Quantity",
    "Previous_Log_Change",
    "Previous_Growth",
    "Year"
]

model_df = long_df[
    long_df["Year"] >= 2021
].copy()

model_df = model_df.dropna(
    subset=FEATURE_COLUMNS + ["Quantity_Kg"]
)

# ================================================================
# 7. CHRONOLOGICAL TRAIN / TEST SPLIT
# ================================================================

train_df = model_df[
    model_df["Year"] <= 2022
].copy()

test_df = model_df[
    model_df["Year"] == 2023
].copy()

if train_df.empty or test_df.empty:
    raise ValueError(
        "Training or testing data is empty."
    )

# ================================================================
# 8. CREATE RELATIVE SEIZURE-VOLUME CLASSES
# ================================================================

# Fit thresholds using training observations only.

LOW_THRESHOLD, HIGH_THRESHOLD = np.quantile(
    train_df["Quantity_Kg"],
    [1/3, 2/3]
)

if LOW_THRESHOLD >= HIGH_THRESHOLD:
    raise ValueError(
        "Unable to create three distinct volume classes."
    )


def classify_risk(quantity):

    if quantity <= LOW_THRESHOLD:
        return 0

    elif quantity <= HIGH_THRESHOLD:
        return 1

    else:
        return 2


train_df["Risk_Class"] = (
    train_df["Quantity_Kg"]
    .apply(classify_risk)
)

test_df["Risk_Class"] = (
    test_df["Quantity_Kg"]
    .apply(classify_risk)
)

X_train = train_df[FEATURE_COLUMNS]

X_test = test_df[FEATURE_COLUMNS]

y_train = train_df["Risk_Class"]

y_test = test_df["Risk_Class"]

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nClassification thresholds:")
print("Low <=", LOW_THRESHOLD)
print("Medium <=", HIGH_THRESHOLD)
print("High >", HIGH_THRESHOLD)

# ================================================================
# 9. DEFINE MACHINE LEARNING MODELS
# ================================================================

models = {

    "Random Forest": Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            RandomForestClassifier(
                n_estimators=150,
                max_depth=5,
                min_samples_leaf=2,
                class_weight="balanced",
                random_state=SEED
            )
        )
    ]),

    "Decision Tree": Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            DecisionTreeClassifier(
                max_depth=4,
                min_samples_leaf=2,
                random_state=SEED
            )
        )
    ]),

    "Logistic Regression": Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            LogisticRegression(
                max_iter=2000,
                class_weight="balanced",
                random_state=SEED
            )
        )
    ]),

    "Gradient Boosting": Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            GradientBoostingClassifier(
                n_estimators=100,
                learning_rate=0.05,
                max_depth=2,
                random_state=SEED
            )
        )
    ])
}

# ================================================================
# 10. TRAIN AND EVALUATE ALL MODELS
# ================================================================

model_results = []

trained_models = {}

model_predictions = {}

for model_name, model in models.items():

    print("\nTraining:", model_name)

    model.fit(
        X_train,
        y_train
    )

    predictions = model.predict(
        X_test
    )

    trained_models[model_name] = model

    model_predictions[model_name] = predictions

    metrics = evaluate_model(
        model_name,
        y_test,
        predictions
    )

    model_results.append(metrics)

    print(
        "Accuracy:",
        round(metrics["Accuracy"], 4)
    )

comparison_df = pd.DataFrame(
    model_results
)

comparison_df = comparison_df.sort_values(
    ["F1_Score", "Accuracy"],
    ascending=False
).reset_index(drop=True)

comparison_df.to_csv(
    BASE_DIR / "model_comparison.csv",
    index=False
)

comparison_df.to_csv(
    BASE_DIR / "accuracy_metrics.csv",
    index=False
)

print("\nMODEL PERFORMANCE")
print(comparison_df.to_string(index=False))

# Choose the highest-scoring model on 2023.
# This is exploratory model selection on a small
# test set, not an independent validation.

BEST_MODEL_NAME = (
    comparison_df.iloc[0]["Model"]
)

BEST_MODEL = trained_models[
    BEST_MODEL_NAME
]

BEST_PREDICTIONS = model_predictions[
    BEST_MODEL_NAME
]

print("\nSelected model:", BEST_MODEL_NAME)

# ================================================================
# 11. ACCURACY GRAPH
# ================================================================

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="Accuracy",
    hue="Model",
    palette="viridis",
    legend=False
)

plt.title(
    "Machine Learning Model Accuracy Comparison",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("Accuracy")
plt.ylim(0, 1.1)

plt.xticks(
    rotation=15,
    ha="right"
)

for container in ax.containers:

    ax.bar_label(
        container,
        fmt="%.3f",
        padding=3
    )

save_show("accuracy_graph.png")

# ================================================================
# 12. MODEL COMPARISON GRAPH
# ================================================================

metrics_to_plot = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1_Score"
]

comparison_long = comparison_df.melt(
    id_vars="Model",
    value_vars=metrics_to_plot,
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=comparison_long,
    x="Model",
    y="Score",
    hue="Metric"
)

plt.title(
    "Comparative Performance of ML Algorithms",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Model")
plt.ylabel("Performance Score")
plt.ylim(0, 1.05)

plt.xticks(
    rotation=15,
    ha="right"
)

plt.legend(
    title="Evaluation Metric",
    loc="upper right"
)

save_show("comparison_graph.png")

# ================================================================
# 13. CORRELATION HEATMAP
# ================================================================

# Heatmap uses historical features and target
# quantity from the training period only.

heatmap_columns = (
    FEATURE_COLUMNS + ["Quantity_Kg"]
)

correlation_data = train_df[
    heatmap_columns
].copy()

correlation_matrix = correlation_data.corr(
    numeric_only=True
)

correlation_matrix.to_csv(
    BASE_DIR / "correlation_matrix.csv"
)

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    square=True
)

plt.title(
    "Correlation Heatmap of Drug Seizure Features",
    fontsize=14,
    fontweight="bold"
)

save_show("heatmap.png")

# ================================================================
# 14. RESULT CSV
# ================================================================

result_df = test_df[
    [
        "Drug",
        "Year",
        "Quantity_Kg",
        "Previous_Quantity"
    ]
].copy()

result_df["Actual_Class"] = (
    y_test.to_numpy()
)

result_df["Predicted_Class"] = (
    BEST_PREDICTIONS
)

result_df["Actual_Risk"] = [
    CLASS_NAMES[int(value)]
    for value in result_df["Actual_Class"]
]

result_df["Predicted_Risk"] = [
    CLASS_NAMES[int(value)]
    for value in result_df["Predicted_Class"]
]

result_df["Correct_Prediction"] = (
    result_df["Actual_Class"]
    ==
    result_df["Predicted_Class"]
)

result_df.to_csv(
    BASE_DIR / "result.csv",
    index=False
)

print("\nRESULT CSV")
print(result_df.head(10))

# ================================================================
# 15. RESULT GRAPH
# ================================================================

result_plot = result_df.sort_values(
    "Drug"
).reset_index(drop=True)

x_positions = np.arange(
    len(result_plot)
)

plt.figure(figsize=(15, 7))

plt.plot(
    x_positions,
    result_plot["Actual_Class"],
    marker="o",
    linewidth=2,
    label="Actual Class"
)

plt.plot(
    x_positions,
    result_plot["Predicted_Class"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Predicted Class"
)

plt.title(
    "Actual vs Predicted Drug Seizure Categories (2023)",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Drug Category")
plt.ylabel("Seizure-Volume Class")

plt.xticks(
    x_positions,
    result_plot["Drug"],
    rotation=75,
    ha="right",
    fontsize=8
)

plt.yticks(
    [0, 1, 2],
    CLASS_NAMES
)

plt.legend()
plt.grid(alpha=0.3)

save_show("result_graph.png")

# ================================================================
# 16. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    BEST_PREDICTIONS,
    labels=[0, 1, 2]
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES
)

plt.title(
    f"Confusion Matrix - {BEST_MODEL_NAME}"
)

plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")

save_show("confusion_matrix.png")

# ================================================================
# 17. GENERATE EXPERIMENTAL 2024 PREDICTIONS
# ================================================================

# Construct 2024 features using 2023 and 2022
# historical quantities. No 2024 observations
# are available in the dataset.

future_df = df[
    ["Drug", "2022", "2023"]
].copy()

future_df = future_df.rename(
    columns={
        "2022": "Two_Years_Ago",
        "2023": "Previous_Quantity"
    }
)

future_df["Year"] = 2024

future_df["Previous_Log_Quantity"] = (
    np.log1p(
        future_df["Previous_Quantity"]
    )
)

future_df["Previous_Log_Change"] = (
    np.log1p(
        future_df["Previous_Quantity"]
    )
    -
    np.log1p(
        future_df["Two_Years_Ago"]
    )
)

future_df["Previous_Growth"] = (
    (
        future_df["Previous_Quantity"]
        -
        future_df["Two_Years_Ago"]
    )
    /
    (
        future_df["Two_Years_Ago"] + 1
    )
)

X_future = future_df[
    FEATURE_COLUMNS
]

# Refit selected model on all labeled years,
# including 2023, before forecasting 2024.
# Thresholds remain fixed from earlier training.

final_model = trained_models[
    BEST_MODEL_NAME
]

final_model.fit(
    model_df[FEATURE_COLUMNS],
    model_df["Quantity_Kg"].apply(classify_risk)
)

future_predictions = final_model.predict(
    X_future
)

future_probabilities = (
    final_model.predict_proba(
        X_future
    )
)

prediction_df = future_df[
    [
        "Drug",
        "Year",
        "Previous_Quantity"
    ]
].copy()

prediction_df["Predicted_Class"] = (
    future_predictions
)

prediction_df["Predicted_Risk"] = [
    CLASS_NAMES[int(value)]
    for value in future_predictions
]

# Class probabilities are relative to the
# fitted model, not calibrated risk probabilities.

for i, class_id in enumerate(
    final_model.classes_
):

    prediction_df[
        f"Probability_{CLASS_NAMES[int(class_id)]}"
    ] = future_probabilities[:, i]

prediction_df["Confidence"] = (
    future_probabilities.max(axis=1)
)

prediction_df["Forecast_Type"] = (
    "Experimental 2024 classification"
)

prediction_df.to_csv(
    BASE_DIR / "prediction.csv",
    index=False
)

print("\n2024 EXPERIMENTAL PREDICTIONS")
print(prediction_df.head(10))

# ================================================================
# 18. PREDICTION GRAPH
# ================================================================

prediction_plot = prediction_df.sort_values(
    "Drug"
).reset_index(drop=True)

x_positions = np.arange(
    len(prediction_plot)
)

plt.figure(figsize=(15, 7))

bars = plt.bar(
    x_positions,
    prediction_plot["Predicted_Class"],
    color=[
        "#2ca25f" if value == 0
        else "#feb24c" if value == 1
        else "#de2d26"
        for value in prediction_plot["Predicted_Class"]
    ]
)

plt.title(
    "Experimental Drug Seizure Category Predictions for 2024",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Drug Category")
plt.ylabel("Predicted Seizure-Volume Class")

plt.xticks(
    x_positions,
    prediction_plot["Drug"],
    rotation=75,
    ha="right",
    fontsize=8
)

plt.yticks(
    [0, 1, 2],
    CLASS_NAMES
)

plt.ylim(-0.1, 2.5)

for bar, label in zip(
    bars,
    prediction_plot["Predicted_Risk"]
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.05,
        label,
        ha="center",
        va="bottom",
        fontsize=7,
        rotation=90
    )

save_show("prediction_graph.png")

# ================================================================
# 19. HISTORICAL YEARLY TREND
# ================================================================

yearly_summary = (
    long_df.groupby("Year")["Quantity_Kg"]
    .sum()
    .reset_index()
)

yearly_summary.to_csv(
    BASE_DIR / "yearly_seizure_summary.csv",
    index=False
)

plt.figure(figsize=(10, 6))

plt.plot(
    yearly_summary["Year"],
    yearly_summary["Quantity_Kg"],
    marker="o",
    linewidth=2.5
)

plt.title(
    "Historical Drug Seizure Quantities (2019-2023)",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Year")
plt.ylabel("Total Seizure Quantity (Kg)")

plt.xticks(
    yearly_summary["Year"]
)

plt.grid(alpha=0.3)

save_show("yearly_trend_graph.png")

# ================================================================
# 20. SAVE TRAINED MODEL
# ================================================================

joblib.dump(
    {
        "model": final_model,
        "features": FEATURE_COLUMNS,
        "class_names": CLASS_NAMES,
        "low_threshold_kg": float(LOW_THRESHOLD),
        "high_threshold_kg": float(HIGH_THRESHOLD),
        "model_name": BEST_MODEL_NAME
    },
    BASE_DIR / "visualization_best_model.pkl"
)

# ================================================================
# 21. FINAL MODEL REPORT
# ================================================================

print("\n" + "=" * 75)
print("FINAL CLASSIFICATION REPORT")
print("=" * 75)

print(
    classification_report(
        y_test,
        BEST_PREDICTIONS,
        labels=[0, 1, 2],
        target_names=CLASS_NAMES,
        zero_division=0
    )
)

print("\nBest model:", BEST_MODEL_NAME)

print(
    "Accuracy:",
    accuracy_score(
        y_test,
        BEST_PREDICTIONS
    )
)

print(
    "F1 Score:",
    f1_score(
        y_test,
        BEST_PREDICTIONS,
        average="weighted",
        zero_division=0
    )
)

# ================================================================
# 22. VERIFY ALL GENERATED FILES
# ================================================================

output_files = [
    "accuracy_metrics.csv",
    "accuracy_graph.png",
    "model_comparison.csv",
    "comparison_graph.png",
    "correlation_matrix.csv",
    "heatmap.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "confusion_matrix.png",
    "yearly_seizure_summary.csv",
    "yearly_trend_graph.png",
    "visualization_best_model.pkl"
]

print("\n" + "=" * 75)
print("OUTPUT FILE VERIFICATION")
print("=" * 75)

for filename in output_files:

    filepath = BASE_DIR / filename

    if filepath.exists():

        print(
            "[SAVED]",
            filename
        )

    else:

        print(
            "[MISSING]",
            filename
        )

print("\n" + "=" * 75)
print("PROJECT EXECUTION COMPLETED")
print("=" * 75)

print("\nAll results saved in:")
print(BASE_DIR)
```
