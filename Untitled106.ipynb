```python
import warnings
warnings.filterwarnings("ignore")

import json
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ================================================================
# 1. CONFIGURATION
# ================================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Drug Seizure Trend Analysis and Risk Classification Using Machine Learning"
)

CSV_PATH = (
    BASE_DIR / "RS_Session_266_AU_242_C_iii.csv"
)

BASE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

rng = np.random.default_rng(SEED)

sns.set_theme(
    style="whitegrid",
    palette="deep"
)

YEARS = [
    "2019",
    "2020",
    "2021",
    "2022",
    "2023"
]

CLASS_NAMES = [
    "Low",
    "Medium",
    "High"
]

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

print("=" * 75)
print("PSO-BASED DRUG SEIZURE CLASSIFICATION")
print("=" * 75)

# ================================================================
# 2. HELPER FUNCTIONS
# ================================================================

def save_show(filename):

    if not filename.startswith("pso_"):
        raise ValueError(
            "Output filenames must start with pso_"
        )

    filepath = BASE_DIR / filename

    plt.tight_layout()

    plt.savefig(
        filepath,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print("[SAVED]", filename)


def evaluate_model(
    model_name,
    actual,
    predicted
):

    return {
        "Model": model_name,

        "Accuracy": float(
            accuracy_score(
                actual,
                predicted
            )
        ),

        "Precision": float(
            precision_score(
                actual,
                predicted,
                average="weighted",
                zero_division=0
            )
        ),

        "Recall": float(
            recall_score(
                actual,
                predicted,
                average="weighted",
                zero_division=0
            )
        ),

        "F1_Score": float(
            f1_score(
                actual,
                predicted,
                average="weighted",
                zero_division=0
            )
        )
    }


# ================================================================
# 3. LOAD DATASET
# ================================================================

df = pd.read_csv(CSV_PATH)

df.columns = (
    df.columns.astype(str).str.strip()
)

DRUG_COL = "Drug (In Kg) / Year"
SERIAL_COL = "Sl. No."

required_columns = (
    [DRUG_COL, SERIAL_COL] + YEARS
)

missing_columns = [
    col
    for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

print("\nOriginal dataset shape:", df.shape)

# ================================================================
# 4. DATA CLEANING
# ================================================================

df[SERIAL_COL] = pd.to_numeric(
    df[SERIAL_COL],
    errors="coerce"
)

# Remove total and summary rows.
df = df.dropna(
    subset=[SERIAL_COL]
).copy()

df["Drug"] = (
    df[DRUG_COL]
    .astype(str)
    .str.strip()
)

# Keep kilogram-based quantities only.
# Do not combine kg, tablets, ampules, or bottles.

exclude_pattern = (
    r"\(In No\.\)|"
    r"\(In Ampules\)|"
    r"\(In Bottles\)"
)

df = df[
    ~df["Drug"].str.contains(
        exclude_pattern,
        case=False,
        regex=True,
        na=False
    )
].copy()

for year in YEARS:

    df[year] = pd.to_numeric(
        df[year],
        errors="coerce"
    )

df = df.dropna(
    subset=YEARS
).copy()

df[YEARS] = df[YEARS].clip(
    lower=0
)

df = df.drop_duplicates(
    subset=["Drug"]
).reset_index(drop=True)

if df.empty:
    raise ValueError(
        "No usable drug seizure records found."
    )

df.to_csv(
    BASE_DIR / "pso_cleaned_dataset.csv",
    index=False
)

print("\nCleaned dataset shape:", df.shape)

# ================================================================
# 5. WIDE TO LONG FORMAT
# ================================================================

long_df = df.melt(
    id_vars=["Drug"],
    value_vars=YEARS,
    var_name="Year",
    value_name="Quantity_Kg"
)

long_df["Year"] = (
    long_df["Year"].astype(int)
)

long_df = long_df.sort_values(
    ["Drug", "Year"]
).reset_index(drop=True)

long_df.to_csv(
    BASE_DIR / "pso_long_format.csv",
    index=False
)

# ================================================================
# 6. FEATURE ENGINEERING
# ================================================================

long_df["Previous_Quantity"] = (
    long_df.groupby("Drug")["Quantity_Kg"]
    .shift(1)
)

long_df["Two_Years_Ago"] = (
    long_df.groupby("Drug")["Quantity_Kg"]
    .shift(2)
)

long_df["Previous_Log_Quantity"] = (
    np.log1p(
        long_df["Previous_Quantity"]
    )
)

long_df["Previous_Log_Change"] = (
    np.log1p(
        long_df["Previous_Quantity"]
    )
    -
    np.log1p(
        long_df["Two_Years_Ago"]
    )
)

long_df["Previous_Growth"] = (
    (
        long_df["Previous_Quantity"]
        -
        long_df["Two_Years_Ago"]
    )
    /
    (
        long_df["Two_Years_Ago"] + 1
    )
)

FEATURES = [
    "Previous_Log_Quantity",
    "Previous_Log_Change",
    "Previous_Growth"
]

model_df = long_df[
    long_df["Year"] >= 2021
].dropna(
    subset=FEATURES + ["Quantity_Kg"]
).copy()

# ================================================================
# 7. TRAIN / TEST SPLIT
# ================================================================

train_df = model_df[
    model_df["Year"] <= 2022
].copy()

test_df = model_df[
    model_df["Year"] == 2023
].copy()

if train_df.empty or test_df.empty:
    raise ValueError(
        "Training or testing dataset is empty."
    )

# Classification thresholds from training data.
LOW_THRESHOLD, HIGH_THRESHOLD = np.quantile(
    train_df["Quantity_Kg"],
    [1/3, 2/3]
)

if LOW_THRESHOLD >= HIGH_THRESHOLD:
    raise ValueError(
        "Unable to create distinct classification thresholds."
    )


def classify_quantity(quantity):

    if quantity <= LOW_THRESHOLD:
        return 0

    elif quantity <= HIGH_THRESHOLD:
        return 1

    else:
        return 2


train_df["Risk_Class"] = (
    train_df["Quantity_Kg"]
    .apply(classify_quantity)
)

test_df["Risk_Class"] = (
    test_df["Quantity_Kg"]
    .apply(classify_quantity)
)

model_df["Risk_Class"] = (
    model_df["Quantity_Kg"]
    .apply(classify_quantity)
)

X_train = train_df[FEATURES]
y_train = train_df["Risk_Class"]

X_test = test_df[FEATURES]
y_test = test_df["Risk_Class"]

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nClassification thresholds:")
print("Low <=", LOW_THRESHOLD)
print("Medium <=", HIGH_THRESHOLD)
print("High >", HIGH_THRESHOLD)

# ================================================================
# 8. PSO OPTIMIZATION CONFIGURATION
# ================================================================

# Particle representation:
#
# [0] Feature 1 selection
# [1] Feature 2 selection
# [2] Feature 3 selection
# [3] Number of estimators
# [4] Maximum tree depth
# [5] Minimum samples per leaf
#
# Feature selected when particle position >= 0.5.

N_PARTICLES = 12
N_ITERATIONS = 20

INERTIA = 0.7
COGNITIVE = 1.5
SOCIAL = 1.5

DIMENSIONS = len(FEATURES) + 3

LOWER_BOUNDS = np.array(
    [0, 0, 0, 40, 2, 1],
    dtype=float
)

UPPER_BOUNDS = np.array(
    [1, 1, 1, 180, 10, 6],
    dtype=float
)

# ================================================================
# 9. PSO VALIDATION DATA
# ================================================================

# 2021: Optimization training
# 2022: Optimization validation
# 2023: Final untouched test year

pso_train_df = train_df[
    train_df["Year"] == 2021
].copy()

pso_val_df = train_df[
    train_df["Year"] == 2022
].copy()

if pso_train_df.empty or pso_val_df.empty:
    raise ValueError(
        "Insufficient PSO validation observations."
    )

X_pso_train = pso_train_df[FEATURES]
y_pso_train = pso_train_df["Risk_Class"]

X_pso_val = pso_val_df[FEATURES]
y_pso_val = pso_val_df["Risk_Class"]

# ================================================================
# 10. PARTICLE DECODING
# ================================================================

def decode_particle(position):

    feature_mask = (
        position[:len(FEATURES)] >= 0.5
    )

    if not feature_mask.any():

        feature_mask[
            np.argmax(
                position[:len(FEATURES)]
            )
        ] = True

    selected = [
        feature
        for feature, flag in zip(
            FEATURES,
            feature_mask
        )
        if flag
    ]

    n_estimators = int(
        np.clip(
            round(position[3]),
            40,
            180
        )
    )

    max_depth = int(
        np.clip(
            round(position[4]),
            2,
            10
        )
    )

    min_samples_leaf = int(
        np.clip(
            round(position[5]),
            1,
            6
        )
    )

    return {
        "features": selected,
        "n_estimators": n_estimators,
        "max_depth": max_depth,
        "min_samples_leaf": min_samples_leaf
    }


def build_pso_model(parameters):

    return RandomForestClassifier(
        n_estimators=parameters["n_estimators"],
        max_depth=parameters["max_depth"],
        min_samples_leaf=parameters["min_samples_leaf"],
        class_weight="balanced",
        random_state=SEED,
        n_jobs=-1
    )

# ================================================================
# 11. PSO FITNESS FUNCTION
# ================================================================

fitness_cache = {}


def fitness_function(position):

    parameters = decode_particle(position)

    selected_features = parameters["features"]

    key = (
        tuple(selected_features),
        parameters["n_estimators"],
        parameters["max_depth"],
        parameters["min_samples_leaf"]
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = build_pso_model(
        parameters
    )

    model.fit(
        X_pso_train[selected_features],
        y_pso_train
    )

    predictions = model.predict(
        X_pso_val[selected_features]
    )

    f1 = f1_score(
        y_pso_val,
        predictions,
        average="macro",
        zero_division=0
    )

    feature_penalty = (
        0.005 * len(selected_features)
    )

    # Minimize negative classification fitness.
    cost = -(f1 - feature_penalty)

    fitness_cache[key] = float(cost)

    return float(cost)

# ================================================================
# 12. INITIALIZE PARTICLE SWARM
# ================================================================

positions = rng.uniform(
    LOWER_BOUNDS,
    UPPER_BOUNDS,
    size=(N_PARTICLES, DIMENSIONS)
)

velocities = np.zeros_like(
    positions
)

personal_best_positions = (
    positions.copy()
)

personal_best_scores = np.array([
    fitness_function(position)
    for position in positions
])

global_best_index = np.argmin(
    personal_best_scores
)

global_best_position = (
    personal_best_positions[
        global_best_index
    ].copy()
)

global_best_score = float(
    personal_best_scores[
        global_best_index
    ]
)

optimization_history = []

# ================================================================
# 13. RUN PARTICLE SWARM OPTIMIZATION
# ================================================================

print("\nStarting PSO optimization...")

for iteration in range(N_ITERATIONS):

    r1 = rng.random(
        size=(N_PARTICLES, DIMENSIONS)
    )

    r2 = rng.random(
        size=(N_PARTICLES, DIMENSIONS)
    )

    velocities = (
        INERTIA * velocities
        +
        COGNITIVE * r1 * (
            personal_best_positions - positions
        )
        +
        SOCIAL * r2 * (
            global_best_position - positions
        )
    )

    positions = positions + velocities

    positions = np.clip(
        positions,
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )

    current_scores = np.array([
        fitness_function(position)
        for position in positions
    ])

    improved = (
        current_scores < personal_best_scores
    )

    personal_best_scores[improved] = (
        current_scores[improved]
    )

    personal_best_positions[improved] = (
        positions[improved].copy()
    )

    best_index = np.argmin(
        personal_best_scores
    )

    if (
        personal_best_scores[best_index]
        <
        global_best_score
    ):

        global_best_score = float(
            personal_best_scores[best_index]
        )

        global_best_position = (
            personal_best_positions[
                best_index
            ].copy()
        )

    optimization_history.append({
        "Iteration": iteration + 1,
        "Best_Fitness": -global_best_score,
        "Best_Cost": global_best_score
    })

    print(
        f"Iteration {iteration + 1}/{N_ITERATIONS}"
        f" | Best Fitness: {-global_best_score:.4f}"
    )

BEST_PARAMETERS = decode_particle(
    global_best_position
)

BEST_FEATURES = (
    BEST_PARAMETERS["features"]
)

print("\nPSO optimization completed.")

print("\nBest parameters:")
print(BEST_PARAMETERS)

print("\nSelected features:")
print(BEST_FEATURES)

# ================================================================
# 14. SAVE PSO OPTIMIZATION HISTORY
# ================================================================

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    BASE_DIR / "pso_optimization_history.csv",
    index=False
)

plt.figure(figsize=(10, 6))

plt.plot(
    history_df["Iteration"],
    history_df["Best_Fitness"],
    marker="o",
    linewidth=2,
    label="Global Best Fitness"
)

plt.title(
    "Particle Swarm Optimization Convergence"
)

plt.xlabel("PSO Iteration")
plt.ylabel("Best Fitness Score")

plt.legend()
plt.grid(alpha=0.3)

save_show("pso_optimization_graph.png")

# ================================================================
# 15. TRAIN OPTIMIZED PSO RANDOM FOREST
# ================================================================

pso_rf = build_pso_model(
    BEST_PARAMETERS
)

pso_rf.fit(
    X_train[BEST_FEATURES],
    y_train
)

pso_predictions = pso_rf.predict(
    X_test[BEST_FEATURES]
)

# ================================================================
# 16. TRAIN BASELINE RANDOM FOREST
# ================================================================

baseline_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=SEED,
    class_weight="balanced",
    n_jobs=-1
)

baseline_rf.fit(
    X_train,
    y_train
)

baseline_predictions = baseline_rf.predict(
    X_test
)

# ================================================================
# 17. TRAIN DECISION TREE
# ================================================================

dt_model = DecisionTreeClassifier(
    max_depth=4,
    random_state=SEED
)

dt_model.fit(
    X_train,
    y_train
)

dt_predictions = dt_model.predict(
    X_test
)

# ================================================================
# 18. TRAIN LOGISTIC REGRESSION
# ================================================================

lr_model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "model",
        LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=SEED
        )
    )
])

lr_model.fit(
    X_train,
    y_train
)

lr_predictions = lr_model.predict(
    X_test
)

# ================================================================
# 19. MODEL PERFORMANCE COMPARISON
# ================================================================

comparison_df = pd.DataFrame([
    evaluate_model(
        "PSO Random Forest",
        y_test,
        pso_predictions
    ),

    evaluate_model(
        "Baseline Random Forest",
        y_test,
        baseline_predictions
    ),

    evaluate_model(
        "Decision Tree",
        y_test,
        dt_predictions
    ),

    evaluate_model(
        "Logistic Regression",
        y_test,
        lr_predictions
    )
])

comparison_df.to_csv(
    BASE_DIR / "pso_comparison.csv",
    index=False
)

comparison_df.to_csv(
    BASE_DIR / "pso_accuracy_metrics.csv",
    index=False
)

print("\nMODEL COMPARISON")
print(comparison_df.to_string(index=False))

# ================================================================
# 20. ACCURACY GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="Accuracy",
    hue="Model",
    palette="viridis",
    legend=False
)

plt.title(
    "PSO vs Machine Learning Model Accuracy",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Machine Learning Model")
plt.ylabel("Accuracy")

plt.ylim(0, 1.1)

plt.xticks(
    rotation=15,
    ha="right"
)

for container in ax.containers:

    ax.bar_label(
        container,
        fmt="%.3f",
        padding=3
    )

save_show("pso_accuracy_graph.png")

# ================================================================
# 21. MODEL COMPARISON GRAPH
# ================================================================

comparison_long = comparison_df.melt(
    id_vars="Model",
    value_vars=[
        "Accuracy",
        "Precision",
        "Recall",
        "F1_Score"
    ],
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(13, 7))

sns.barplot(
    data=comparison_long,
    x="Model",
    y="Score",
    hue="Metric"
)

plt.title(
    "PSO Comparative Model Performance",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Model")
plt.ylabel("Performance Score")

plt.ylim(0, 1.05)

plt.xticks(
    rotation=15,
    ha="right"
)

plt.legend(
    title="Evaluation Metric"
)

save_show("pso_comparison_graph.png")

# ================================================================
# 22. CORRELATION HEATMAP
# ================================================================

heatmap_columns = list(
    dict.fromkeys(
        BEST_FEATURES + ["Quantity_Kg"]
    )
)

correlation_matrix = train_df[
    heatmap_columns
].corr(
    numeric_only=True
)

correlation_matrix.to_csv(
    BASE_DIR / "pso_correlation_matrix.csv"
)

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    square=True
)

plt.title(
    "PSO Selected Features Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

save_show("pso_heatmap.png")

# ================================================================
# 23. SAVE RESULT CSV
# ================================================================

result_df = test_df[
    [
        "Drug",
        "Year",
        "Quantity_Kg",
        "Previous_Quantity"
    ]
].copy()

result_df["Actual_Class"] = (
    y_test.to_numpy()
)

result_df["PSO_Predicted_Class"] = (
    pso_predictions
)

result_df["Actual_Risk"] = [
    CLASS_NAMES[int(value)]
    for value in y_test
]

result_df["PSO_Predicted_Risk"] = [
    CLASS_NAMES[int(value)]
    for value in pso_predictions
]

result_df["Correct_Prediction"] = (
    result_df["Actual_Class"]
    ==
    result_df["PSO_Predicted_Class"]
)

result_df.to_csv(
    BASE_DIR / "pso_result.csv",
    index=False
)

print("\nPSO CLASSIFICATION RESULTS")
print(result_df.head(10))

# ================================================================
# 24. RESULT GRAPH
# ================================================================

result_plot = result_df.sort_values(
    "Drug"
).reset_index(drop=True)

x_positions = np.arange(
    len(result_plot)
)

plt.figure(figsize=(15, 7))

plt.plot(
    x_positions,
    result_plot["Actual_Class"],
    marker="o",
    linewidth=2,
    label="Actual Class"
)

plt.plot(
    x_positions,
    result_plot["PSO_Predicted_Class"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Predicted Class"
)

plt.xticks(
    x_positions,
    result_plot["Drug"],
    rotation=75,
    ha="right",
    fontsize=8
)

plt.yticks(
    [0, 1, 2],
    CLASS_NAMES
)

plt.title(
    "Actual vs PSO Predicted Classification (2023)",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Drug")
plt.ylabel("Seizure-Volume Category")

plt.legend()
plt.grid(alpha=0.3)

save_show("pso_result_graph.png")

# ================================================================
# 25. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    pso_predictions,
    labels=[0, 1, 2]
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES
)

plt.title(
    "PSO Optimized Random Forest Confusion Matrix"
)

plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")

save_show("pso_confusion_matrix.png")

# ================================================================
# 26. GENERATE EXPERIMENTAL 2024 PREDICTIONS
# ================================================================

future_df = df[
    ["Drug", "2022", "2023"]
].copy()

future_df = future_df.rename(
    columns={
        "2022": "Two_Years_Ago",
        "2023": "Previous_Quantity"
    }
)

future_df["Year"] = 2024

future_df["Previous_Log_Quantity"] = (
    np.log1p(
        future_df["Previous_Quantity"]
    )
)

future_df["Previous_Log_Change"] = (
    np.log1p(
        future_df["Previous_Quantity"]
    )
    -
    np.log1p(
        future_df["Two_Years_Ago"]
    )
)

future_df["Previous_Growth"] = (
    (
        future_df["Previous_Quantity"]
        -
        future_df["Two_Years_Ago"]
    )
    /
    (
        future_df["Two_Years_Ago"] + 1
    )
)

# Refit the optimized model on all labeled years.

final_pso_model = build_pso_model(
    BEST_PARAMETERS
)

final_pso_model.fit(
    model_df[BEST_FEATURES],
    model_df["Risk_Class"]
)

future_predictions = final_pso_model.predict(
    future_df[BEST_FEATURES]
)

future_probabilities = (
    final_pso_model.predict_proba(
        future_df[BEST_FEATURES]
    )
)

prediction_df = future_df[
    [
        "Drug",
        "Year",
        "Previous_Quantity"
    ]
].copy()

prediction_df["PSO_Predicted_Class"] = (
    future_predictions
)

prediction_df["PSO_Predicted_Risk"] = [
    CLASS_NAMES[int(value)]
    for value in future_predictions
]

for i, class_id in enumerate(
    final_pso_model.classes_
):

    prediction_df[
        f"Probability_{CLASS_NAMES[int(class_id)]}"
    ] = future_probabilities[:, i]

prediction_df["Confidence"] = (
    future_probabilities.max(axis=1)
)

prediction_df["Forecast_Type"] = (
    "Experimental PSO 2024 classification"
)

prediction_df.to_csv(
    BASE_DIR / "pso_prediction.csv",
    index=False
)

print("\nPSO 2024 PREDICTIONS")
print(prediction_df.head(10))

# ================================================================
# 27. PREDICTION GRAPH
# ================================================================

prediction_plot = prediction_df.sort_values(
    "Drug"
).reset_index(drop=True)

x_positions = np.arange(
    len(prediction_plot)
)

colors = [
    "#2ca25f" if value == 0
    else "#feb24c" if value == 1
    else "#de2d26"
    for value in prediction_plot["PSO_Predicted_Class"]
]

plt.figure(figsize=(15, 7))

bars = plt.bar(
    x_positions,
    prediction_plot["PSO_Predicted_Class"],
    color=colors
)

plt.xticks(
    x_positions,
    prediction_plot["Drug"],
    rotation=75,
    ha="right",
    fontsize=8
)

plt.yticks(
    [0, 1, 2],
    CLASS_NAMES
)

plt.ylim(-0.1, 2.5)

plt.title(
    "PSO Experimental Drug Seizure Predictions for 2024",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Drug")
plt.ylabel("Predicted Seizure-Volume Category")

save_show("pso_prediction_graph.png")

# ================================================================
# 28. FEATURE IMPORTANCE
# ================================================================

importance_df = pd.DataFrame({
    "Feature": BEST_FEATURES,
    "Importance": final_pso_model.feature_importances_
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

importance_df.to_csv(
    BASE_DIR / "pso_feature_importance.csv",
    index=False
)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature",
    hue="Feature",
    legend=False
)

plt.title(
    "PSO Optimized Feature Importance"
)

save_show("pso_feature_importance_graph.png")

# ================================================================
# 29. SAVE PSO MODEL
# ================================================================

joblib.dump(
    {
        "model": final_pso_model,
        "selected_features": BEST_FEATURES,
        "best_parameters": BEST_PARAMETERS,
        "class_names": CLASS_NAMES,
        "low_threshold_kg": float(LOW_THRESHOLD),
        "high_threshold_kg": float(HIGH_THRESHOLD),
        "optimization_method": "Particle Swarm Optimization"
    },
    BASE_DIR / "pso_model.pkl"
)

# ================================================================
# 30. SAVE PSO METRICS JSON
# ================================================================

pso_metrics = {
    "project": (
        "AI-Based Drug Seizure Trend Analysis "
        "and Risk Classification"
    ),

    "optimization": "Particle Swarm Optimization",

    "n_particles": N_PARTICLES,
    "n_iterations": N_ITERATIONS,

    "inertia": INERTIA,
    "cognitive_coefficient": COGNITIVE,
    "social_coefficient": SOCIAL,

    "best_fitness": float(
        -global_best_score
    ),

    "selected_features": BEST_FEATURES,

    "best_parameters": BEST_PARAMETERS,

    "training_samples": int(
        len(X_train)
    ),

    "testing_samples": int(
        len(X_test)
    ),

    "training_years": [2021, 2022],
    "testing_year": 2023,

    "classification_thresholds_kg": {
        "low": float(LOW_THRESHOLD),
        "medium": float(HIGH_THRESHOLD)
    },

    "model_comparison": {
        row["Model"]: {
            "accuracy": float(row["Accuracy"]),
            "precision": float(row["Precision"]),
            "recall": float(row["Recall"]),
            "f1_score": float(row["F1_Score"])
        }
        for _, row in comparison_df.iterrows()
    },

    "confusion_matrix": cm.tolist(),

    "evaluation_note": (
        "Exploratory classification with a small "
        "historical dataset. The categories describe "
        "relative seizure volume, not actual crime risk."
    )
}

with open(
    BASE_DIR / "pso_metrics.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        pso_metrics,
        file,
        indent=4
    )

# ================================================================
# 31. SAVE PSO CONFIGURATION YAML
# ================================================================

pso_config = {
    "project": (
        "AI-Based Drug Seizure Trend Analysis "
        "and Risk Classification"
    ),

    "dataset": CSV_PATH.name,

    "optimization": "Particle Swarm Optimization",

    "n_particles": N_PARTICLES,
    "n_iterations": N_ITERATIONS,

    "inertia": INERTIA,
    "cognitive_coefficient": COGNITIVE,
    "social_coefficient": SOCIAL,

    "selected_features": BEST_FEATURES,
    "best_parameters": BEST_PARAMETERS,

    "target": "Risk_Class",

    "class_names": CLASS_NAMES,

    "training_years": [2021, 2022],
    "testing_year": 2023,
    "forecast_year": 2024,

    "model_file": "pso_model.pkl"
}

with open(
    BASE_DIR / "pso_config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        pso_config,
        file,
        sort_keys=False
    )

# ================================================================
# 32. FINAL CLASSIFICATION REPORT
# ================================================================

print("\n" + "=" * 75)
print("PSO CLASSIFICATION REPORT")
print("=" * 75)

print(
    classification_report(
        y_test,
        pso_predictions,
        labels=[0, 1, 2],
        target_names=CLASS_NAMES,
        zero_division=0
    )
)

print(
    "\nPSO Accuracy:",
    accuracy_score(
        y_test,
        pso_predictions
    )
)

print(
    "\nPSO F1 Score:",
    f1_score(
        y_test,
        pso_predictions,
        average="weighted",
        zero_division=0
    )
)

# ================================================================
# 33. VERIFY ALL OUTPUT FILES
# ================================================================

output_files = [
    "pso_cleaned_dataset.csv",
    "pso_long_format.csv",
    "pso_optimization_history.csv",
    "pso_optimization_graph.png",
    "pso_accuracy_metrics.csv",
    "pso_accuracy_graph.png",
    "pso_comparison.csv",
    "pso_comparison_graph.png",
    "pso_correlation_matrix.csv",
    "pso_heatmap.png",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_confusion_matrix.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_feature_importance.csv",
    "pso_feature_importance_graph.png",
    "pso_model.pkl",
    "pso_metrics.json",
    "pso_config.yaml"
]

print("\n" + "=" * 75)
print("PSO OUTPUT VERIFICATION")
print("=" * 75)

for filename in output_files:

    filepath = BASE_DIR / filename

    if filepath.exists():

        print(
            "[SAVED]",
            filename
        )

    else:

        print(
            "[MISSING]",
            filename
        )

print("\n" + "=" * 75)
print("PSO PROJECT COMPLETED SUCCESSFULLY")
print("=" * 75)

print("\nAll files saved in:")
print(BASE_DIR)
```
